Zadanie 1 \
State + prosty węzeł
Podmień pytanie przypisane do klucza "question" i uruchom ponizszy kod.

```python
# Zainstaluj bezpośrednie zależności używane w tym notebooku.
# Pomijane, gdy brak pip, np. w środowisku `uv sync`, które ma je już zainstalowane (zob. README).
import importlib.util
if importlib.util.find_spec("pip"):
    %pip install -qU langgraph
```

```python
from typing import TypedDict, Optional

class State(TypedDict, total=False):
    question: str
    answer: Optional[str]

def echo_node(state: State) -> State:
    new_state = dict(state)
    if "answer" not in new_state:
        new_state["answer"] = f"ECHO: {state.get('question','')}"
    return new_state

assert callable(echo_node)

state = State({"question": "Czym jest LangGraph?"}) # <- podmień na swój tekst
echo_node(state)
```

Zadanie 2 \
Graf: START → echo → END \
Uzupełnij "____" (klucz i pytanie). Następnie uruchom kod prostego grafu.

```python
try:
    from langgraph.graph import StateGraph, START, END
except Exception as e:
    print("LangGraph nie jest zainstalowany (demo).", e)
    StateGraph = object
    START="__START__"
    END="__END__"

builder = StateGraph(State)  # TODO: typ stanu
try:
    builder.add_node("echo", echo_node)
    builder.add_edge(START, "echo")
    builder.add_edge("echo", END)
except Exception as e:
    print("Szkic dodawania węzłów/krawędzi:", e)

try:
    graph = builder.compile()
    assert graph is not None
except Exception:
    pass

graph.invoke({"____": "____"})
```
